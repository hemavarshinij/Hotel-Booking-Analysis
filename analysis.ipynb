Adults + Children + Babies

```python
df['Total_Guests'] = (
    df['adults'] +
    df['children'] +
    df['babies']
)
```

```python
df[['adults', 'children', 'babies', 'Total_Guests']].head()
```

Weekend Nights + Week Nights

```python
df['Stay_Nights'] = (
    df['stays_in_weekend_nights'] +
    df['stays_in_week_nights']
)
```

```python
df[
    [
        'stays_in_weekend_nights',
        'stays_in_week_nights',
        'Stay_Nights'
    ]
].head()
```

is_canceled = 0 → Completed
is_canceled = 1 → Cancelled

```python
df['Booking_Status'] = df['is_canceled'].map({
    0: 'Completed',
    1: 'Cancelled'
})
```

```python
df[['is_canceled', 'Booking_Status']].head()
```

### BASIC ANALYSIS

```python
#TOTAL BOOKINGS

total_bookings = len(df)

print("Total Bookings:", total_bookings)
```

```python
#CANCELLED BOOKINGS

total_cancelled = (
    df['Booking_Status'] == 'Cancelled'
).sum()

print("Total Cancelled Bookings:", total_cancelled)
```

```python
#CANCELLATION RATE

cancellation_rate = (
    total_cancelled / total_bookings
) * 100

print("Cancellation Rate:",
      round(cancellation_rate, 2), "%")
```

```python
#AVERAGE DAILY RATE

average_adr = df['adr'].mean()

print("Average ADR:",
      round(average_adr, 2))
```

```python
#AVERAGE STAY DURATION

average_stay = df['Stay_Nights'].mean()

print("Average Stay Duration:",
      round(average_stay, 2),
      "nights")
```

```python
#CITY HOTEL BOOKINGS

city_bookings = (
    df['hotel'] == 'City Hotel'
).sum()

print("City Hotel Bookings:",
      city_bookings)
```

```python
#RESORT HOTEL BOOKINGS

resort_bookings = (
    df['hotel'] == 'Resort Hotel'
).sum()

print("Resort Hotel Bookings:",
      resort_bookings)
```

```python
#CREATING A TABLE USING ALL THE ANALYSIS DONE(REFERENCE)

summary = pd.DataFrame({
    'Metric': [
        'Total Bookings',
        'Cancelled Bookings',
        'Cancellation Rate',
        'Average ADR',
        'Average Stay Duration',
        'City Hotel Bookings',
        'Resort Hotel Bookings'
    ],

    'Value': [
        total_bookings,
        total_cancelled,
        round(cancellation_rate, 2),
        round(average_adr, 2),
        round(average_stay, 2),
        city_bookings,
        resort_bookings
    ]
})

summary
```
