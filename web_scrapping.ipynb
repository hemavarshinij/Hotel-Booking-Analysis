# WEB SCRAPPING

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
Url="https://en.wikipedia.org/wiki/World_Tourism_rankings"
headers = {
    "User-Agent": "Mozilla/5.0"
}
```

```python
response = requests.get(Url,headers=headers)
```

```python
response.status_code
```

### a. find() or find_all()

```python
soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table")
rows = table.find_all("tr")
```

```python
data =[]
```

### LOOPS


```python
for row in rows[1:11]:   
    columns = row.find_all(["th", "td"])

    if len(columns) >= 2:
        country = columns[0].get_text(strip=True)
        tourist_arrivals = columns[1].get_text(strip=True)

        data.append({
            "Country": country,
            "Tourism Rank or Visitor Information": tourist_arrivals
        })
```

### PANDAS DATAFRAME

```python
df_tourism = pd.DataFrame(data)
```

```python
print(df_tourism)
```

```python
df_tourism.to_csv("tourism_data.csv", index=False)

print("\nFile saved successfully as tourism_data.csv")
```

### COMPLETE SCRAPING

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
Url="https://en.wikipedia.org/wiki/World_Tourism_rankings"
headers = {
    "User-Agent": "Mozilla/5.0"
}
response = requests.get(Url,headers=headers)
response.status_code
soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table")
rows = table.find_all("tr")
data =[]
for row in rows[1:11]:   
    columns = row.find_all(["th", "td"])

    if len(columns) >= 2:
        country = columns[0].get_text(strip=True)
        tourist_arrivals = columns[1].get_text(strip=True)

        data.append({
            "Country": country,
            "Tourism Rank or Visitor Information": tourist_arrivals
        })
df_tourism = pd.DataFrame(data)
print(df_tourism)
```
